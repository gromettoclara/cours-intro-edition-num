# Un peu de théorie…

## Point terminologique

`Document :`

- 1 support contenant une ou plusieurs informations (contenu)
- textuel ou non-textuel (iconographique, sonore, audiovisuel, ...)
- simple ou complexe (multimédia)

Un `document physique` existe dans le monde matériel.

Un `document numérique` existe sur un plan matériel (hardware) et immatériel (software)

- contenu encodé sous la forme de suites de 0 et de 1
- matérialisé virtuellement sous la forme d'un fichier
- il a un format qui doit être décodé par la machine pour être lisible par l'humain
- selon le format, il est plus ou moins sujet au traitement numérique

`dématérialisé` ou `numérisé` vs. `nativement numérique`

## Les traitements de texte : histoire 

⌨️ Jusque dans les années 1980, on écrit essentiellement du code.

⌨️ Les premiers logiciels (*Electric Pencil* 1976, *Easy Writer* 1979) visent la documentation du code, pas l’impression. Les premiers éditeurs WYSIWYG (*Bravo* 1974) introduisent la notion de page, mais le rendu imprimé reste limité.

⌨️ Avec *WordStar* (1979) et *MacWrite* (1984), la cible devient clairement les entreprises et le secrétariat. Le traitement de texte devient aussi un outil de productivité et de rationalisation de processus de travail.

⌨️ À la fin des années 1980, Microsoft Word s’impose comme standard, démocratisant le traitement de texte pour la bureautique et la préparation de manuscrits.

## Exemple d'usage -- CTS (*Canonical Text Service*)

📜 protocole standard de citation des textes anciens 

📜 l'accès à ces extraits de textes se fait via un protocole standard (`API` -- *Application Programming Interface*) qui utilise des `requêtes HTTP` ("hacker" des url)

## URI vs. URN vs. URL

`URI` = *Uniform Ressource Indentifier* 

Elle peut être : 

🪪 Une `URN` (*Uniform Ressource Name*)  = identification de la ressource indépendemment de son emplacement 

`urn:cts:greekLit:tlg0012.tlg001.perseus-grc2:1.1-1.10`

`greekLit` : namespace (littérature grecque)

`tlg0012` : textgroup (Homère)

`tlg001` : work (L'Iliade)

`perseus-grc2` : version (édition Perseus en grec)

`1.1-1.10` : passage (livre 1, vers 1 à 10)

📍🗺️ Une `URL` (*Uniform Ressource Locator*) = le chemin vers cette ressource

## L'API

🚚📦 une `API` dite `REST` repose sur le protocole HTTP

🚚📦 on fait une `requête` (une commande) à un serveur

📬 grâce à des structures d'URL standards, le serveur sait quelle "partie de texte" il doit livrer

`http://votre-serveur.com/api/cts?request=GetPassage&urn=urn:cts:greekLit:tlg0059.tlg030.votre-edition:1.1`

# Les formats

## Le fichier numérique

### Propriétés élémentaires :

- un `nom`
- un `format` indiqué par son `extension`
- une `taille` (ou poids)

### Des métadonnées (informations sur le contexte et les modalités de création)

- `date` de création, de modification
- identité des `créateur-rice(s)`
- `logiciel` de création
- etc...

## Définition

> "[…] un format, c’est quelque chose qui met en forme d’une manière conventionnelle des données destinées à représenter du texte, du son, de l’image, de la vidéo, ou une combinaison des quatre. C’est une sorte de « gabarit » qui met certaines données à des endroits déterminés. Les outils qui vont traiter ce format s’attendent à trouver tel élément à tel endroit, organisé de telle manière, et tel autre à un autre endroit, organisé d’une autre façon." (Boulétreau, Viviane, et Benoît Habert. 2014. « Chapitre 9. Les formats ». In *Pratiques de l’édition numérique*, édité par Marcello Vitali-Rosati et Michael E. Sinatra. Parcours numérique. Presses de l’Université de Montréal. https://doi.org/10.4000/books.pum.329.)

> "[s]tructure définie de données contenues sur un support magnétique ou autre, établie selon des règles qui régissent le stockage, l'affichage, la manipulation, l'impression ou la transmission de ces données." (OQLF, 2001)

**Un format n’est pas…**

- un logiciel
- une entreprise
- une « extension »

L'`extension` d'un fichier

- correspond à des normes (potentiellement infinies)
- nous donne une information sur la nature de son contenu
- indique à la machine comment le traiter
- permet aux `logiciels` (développés ou non par des `entreprises`) de trier les fichiers compatibles

## Différentes catégories de formats

### Format propriétaire

- spécifications techniques sont la propriété (intellectuelle) d'une entité privée
- souvent brevetées, donc utilisation soumise à des limitations

- `opaque` :
    - pas de diffusion
    - le fichier ne peut être traité que par le logiciel (propriétaire) qui le produit
- `publié` :
    - quand même soumis aux limitations du brevet
    - le fichier peut être traité par d'autres logociels
    - risque de perte d'accès aux specs

### Format libre/ouvert

- spécifications techniques publiques et sans restriction d'accès
- le fichier peut être traité par n'importe quel logiciel développé pour le supporter
- pas toujours standardisé (c'est-à-dire pris en main par une communauté)

## Comment naît un format ? 

### Standard de fait

- spécifications techniques existent, sont publiques et adoptées très largement
- mais n'ont pas (encore) fait l'objet d'une standardisation officielle

### Norme

- spécifications techniques ont fait l'objet d'une standardisation officielle
- exemples d'organisation de normalisation : ISO, AFNOR, W3C

## Des choix importants

- créer un `standard` = documenter un format, pour permettre à une communauté d'utilisateur‧ices de s'en emparer pour créer des logiciels pour l'ouvrir ou pour le modifier
- apposer une `licence` = de quelle façon on peut utiliser ce gabarit.
    - lecture uniquement
    - modification  
    - redistribution

## Qu'implique le choix d'un format ?

- interopérabilité
- pérennité

## Différents types de fichiers

### Fichier complexe ou Fichier multimédia :

- fichiers complexes mélangeant plusieurs types de données (textuelles / non-textuelles)
- généralement le résultat de la compression de plusieurs fichiers numériques simples
- ex: .doc, .odt, .pdf

### Fichier exécutable :

- peuvent être exécutés par un programme pour réaliser une action
- ex : .exe, .sh, .app, ...

### Fichier compressé ou *Document Container File* :

- regroupent plusieurs fichiers en un seul
- peuvent être décompressé à l'aide d'un logiciel
- peuvent réduire la taille globale du lot de fichier
- ex : .zip, .tar, .rar, .docx


## Recap -- le document numérique

- plusieurs couches d'appareillage technique sont nécessaires
- le format définit un gabarit public ou non, libre ou non, pour structurer l'information
- le statut d'un format dépend en grande partie du statut des spécifications techniques
- il existe une grande variété de types de fichiers, indiqué par leur extension
- certains fichiers sont des combinaison de plusieurs fichiers

# XML

## XML

🌳 XML est apparu à la fin des années 1990, c'est un sous-ensemble du standard SGML (Standard Generalized Markup Language), comme HTML (Hypertext Markup Language)

🌳 Ces langages utilisent les chevrons (<>) pour former des balises.

🌳 XML signifie *eXtensible Markup Language*

🌳 XML est un langage de balisage générique

  `<balise>contenu</balise>`

🌳 XML est dit `extensible` car on peut créer ses propres noms de balise


## La syntaxe du XML

🌳 une balise XML est écrite entre chevrons (< nom_de_la_balise >) et entoure généralement un contenu

🌳 une balise ouverte (`<slide>`) doit donc être fermée (`</slide>`)

🌳 une balise peut-être vide (elle n'encadre pas un contenu) : `<nothing></nothing>`, ce qui peut aussi s'écrire : `<nothing/>`

🌳 les balises peuvent être imbriquées ([modèle OHCO](https://www.cip.ifi.lmu.de/~langeh/test/1990%20-%20DeRose%20-%20What%20is%20Text,%20really%3F.pdf))

🌳 un commentaire est écrit entre `<!-- ... -->` ;

🌳 une balise peut avoir des attributs qui possèdent des valeurs ;

`<balise attribut="valeur">contenu</balise>`

## XML : un exemple

```
<slide>
    <titre>Qu'est-ce que XML ?</titre>
    <para>
        <line n="001">🌳 XML signifie <anglais>eXtensible Markup Language</anglais></line>
        <line n="002">🌳 XML est un <gras>langage</gras> de <gras>balisage</gras> générique. On le dit <gras>extensible</gras> car il propose une grammaire générale pour l'organisation du document, mais on peut créer ses propres noms de balises</line>
    </para>
</slide>
```

- `<slide>` est une balise ou un élément
- "slide" est un nom de balise
- "Qu'est-ce que XML ?" est le contenu de la balise "titre"
- "n" est un attribut de la balise "line"
- "001" est la valeur de l'attribut "n"

## XML et les standards

🌳 XML n'indique pas comment s'appellent les balises

`<paragraph>, <paragraphe>, <para>, <para_ph>, <p>` sont tous des noms de balises valides

🌳 XML ne correspond qu'à la syntaxe

🌳 Les standards définissent le vocabulaire (ensemble de noms de balise) et des règles de compositions :

- XML TEI (*Text Encoding Initiative*) - description de documents textuels (1987)
- XML EAD (*Encoded Archival Description*) - description archivistique (1998)
- XML SVG (*Scalable Vector Graphics*) - descriptions d'images vectorielles (1999)
- XML ALTO (*Analysed Layout and Text Object*) - description de la mise en page et du contenu d'un texte transcrit automatiquement (2004)

🌳 Le schéma peut définir une sous-spécification au sein d'un standard. 

- rajouter des contraintes
- décrire ce qui est dans les éléments

## Concepts 

🌳 le `format` XML désigne donc plusieurs choses :

- un fichier de texte
- un arbre composé d'éléments hiérarchisés : ils doivent être strictement imbriqués les uns dans les autres, comme des poupées russes. 
- un `métalangage` (eXtensible)

🌳 c'est donc un format doublement standardisé, parce qu'il y a à la fois la base XML et une application spécifique de XML

🌳 le langage XML a un rôle central dans le processus de normalisation des formats : ePub, Office Open XML, PDF reposent sur XML et sur la notion d'archive pour regrouper des ressources hétérogènes en un seul fichier

## Outils

🌳 Un fichier XML peut aussi constituer une base de donnée à interroger

🌳 Il existe plusieurs langages qui permettent d'interagir avec XML (centré sur la structure des données, ne dit pas comment afficher) :

- xPath : formuler un chemin à l'intérieur d'un arbre XML
- XSLT : transformer un arbre XML en un autre arbre XML ou en HTML
- xQuery : permet de faire des requêtes dans une base XML

🌳 plusieurs `schémas` = des règles qui définissent quelles balises sont autorisées et pour quel usage

🌳 des moyens de validation qui les accompagnent

**DTD**

```
<!ELEMENT personne (nom, age)>
<!ELEMENT nom (#PCDATA)>
<!ELEMENT age (#PCDATA)>
```

**XML Schema** (aussi appelé XSD)

```
<xs:schema xmlns:xs="http://www.w3.org/2001/XMLSchema">

  <xs:element name="personne">
    <xs:complexType>
      <xs:sequence>
        <xs:element name="nom" type="xs:string"/>
        <xs:element name="age" type="xs:integer"/>
      </xs:sequence>
    </xs:complexType>
  </xs:element>

</xs:schema>

```

**Relax NG**

```
<element name="personne" xmlns="http://relaxng.org/ns/structure/1.0"
         xmlns:xsd="http://www.w3.org/2001/XMLSchema-datatypes">
  <element name="nom">
    <text/>
  </element>
  <element name="age">
    <data type="xsd:integer"/>
  </element>
</element>

```

# Markdown

## Markdown

Ⓜ️ Créé en 2004 par John Gruber et Aaron Swartz

Ⓜ️ Markdown est un langage de balisage léger qui permet d'encoder facilement la structure et certains éléments de mise en forme du texte sans gêner sa lecture

Ⓜ️ Il permet de créer des fichiers HTML ou PDF (c'est rarement le format cible, il fait office de pivot)

Ⓜ️ Même non compilé, un texte Markdown est facile à lire car les annotation sont légères et limitées

🤫 On peut utiliser Markdown ou un système similaire dans Whatsapp et Messenger
