# Édition / éditorialisation / SSP

*L'édition, acte de productif de sens, trouve dans l'éditorialisation une reformulation de ses fonctions essentielles (stabilisation, critique, agrégation) adaptée aux dynamiques ouvertes, collectives et processuelles de l'espace numérique.*

## L'édition

Édition = un acte, une production (*edit* + *publish*)

- acte sur le texte 
- acte documentaire 
- acte politique

La `chaîne` éditoriale = tout ce qui se trouve entre l'établissement du texte et sa publication. 

> Lʼédition est un moyen dʼagir, cʼest une activité politique. Il est primordial de prendre en considération cette dimension et donc la puissance potentielle dʼune structure dʼédition ou des personnes qui la représentent. Quand bien même une telle structure semble petite par la taille de son catalogue ou de son équipe, placer un texte dans lʼespace public a un effet sur cet espace : répercussions médiatiques, réactions populaires, modifications sociales, débats citoyens, etc. (Fauchié, Antoine. 2024. « Fabriquer des éditions, éditer des fabriques : reconfiguration des processus techniques éditoriaux et nouveaux modèles épistémologiques ». Université de Montréal. https://these.quaternum.net/, p.99)

##  L'édition s'institutionnalise autour de la réponse aux besoins de : 

- stabiliser 
- critiquer
- aggréger

des textes et des idées. 

## L'éditorialisation désigne un ensemble de pratique qui répondent à ces mêmes besoins avec les outils numériques

> Je propose de définir l'éditorialisation comme l'ensemble des dynamiques qui produisent l'espace numérique. Ces dynamiques peuvent être comprises comme les interactions d'actions individuelles et collectives avec un environnement numérique. (Vitali Rosati, Marcello. 2016. « Qu’est-ce que l’éditorialisation ? » *Sens public.* https://doi.org/10.7202/1043383ar.)

### Caractéristiques 

- ouverte dans le temps
- ouverte dans l'espace
- collective

### La dimension collective se manifeste à plusieurs niveaux :

- Dans le processus de production et d'écriture (facilité par les outils numériques)
- Dans une dynamique conversationnelle (commentaires, réactions)
- Dans l'indexation collective (en cliquant ou non sur les contenus, on participe à l'architecture de l'information)
- Dans la citation mutuelle (en choisissant de mettre un lien vers un autre contenu)

Des conséquences pratiques : la chaîne éditoriale numérique

## Le *Single Source Publishing* (SSP)

- une source unique → plusieurs formats, pour plusieurs modes de diffusion, et pour des perceptions différentes du même contenu

- séparation contenu / forme (balisage sémantique + conversion)

### ✓ Les avantages 

- cohérence
- pérennité des sources 
- éviter des coûts 
- intervention des métiers au bon endroit 
- redéfinir et redonner sa place au travail d'édition
- cohérence, gain de temps et de ressources

## Le numérique souligne les problématiques, les tensions et les besoins de l'édition, apportant de nouveaux défis

💻 introduire la dimension collaborative

💻 remettre en jeu la fixité des contenus

💻 conserver pourtant la cohérence des sources…

💻 … et ce malgré l'ubiquité des contenus

# Collaborer sur une chaîne éditoriale

## GIT ?!

`git` est un outil de gestion de versions décentralisé créé par Linus Torvald 
(le même que Linux) en 2005 sous licence open-source GPL

## À quoi ça sert le _versioning_ ?

1. **_Sauvegarder des versions_** : Git permet de créer des instantanés du contenu d’un dossier
2. **_Conserver un historique_** : En sauvegardant des versions successives, on a accès à la liste de toutes les modifications
3. **_Créer des états parallèles_** : Possibilité de disposer de plusieurs versions simultanées d’un dossier
4. **_Collaborer sur un projet_** : Les contributions effectuées sur un même fichier peuvent être fusionnées

## Sauvegarder des versions

<figure style="display: flex; align-items: center; justify-content: center;">
  <img src="media/versions.png" alt="logo" style="height: 200px; width: auto;">
</figure>

Avec git, en plus de sauvegarder normalement vos fichiers, vous pouvez décider de **capturer des états donnés** d’un de vos dossiers.

Un dossier dont vous enregistrez des versions se nomme _**`repository`**_.

Une sauvegarde de ce repository est nommée _**`commit`**_.

## Conserver un historique

<figure style="display: flex; align-items: center; justify-content: center;">
  <img src="media/ctrlz.png" alt="logo" style="height: 200px; width: auto;">
</figure>

Chaque _**`commit`**_ conserve en mémoire :

- sa date de création
- les lignes précises où les fichiers ont été édités
- un message expliquant le but des modifications
- la personne qui l’a effectué

En remontant les _commits_ successifs, il est possible de suivre précisément l’évolution d’un _repository_ et de revenir à un état précédent au besoin.

## Créer des états parallèles

<figure style="display: flex; align-items: center; justify-content: center;">
  <img src="media/etat-parallele.png" alt="logo" style="height: 200px; width: auto;">
</figure>

Il est possible de faire coexister différentes versions des fichiers contenus dans un _repository_ :
ces versions sont appelées _**`branches`**_.

Les _**`branches`**_ permettent de faire cohabiter de multiples états d’avancement d’un projet et 
de se lancer dans des modifications sans casser l’existant.

La branche par défaut s’appelle _**`main`**_ (anciennement _**`master`**_).

## Collaborer sur un projet

<figure style="display: flex; align-items: center; justify-content: center;">
  <img src="media/collaborer.png" alt="logo" style="height: 200px; width: auto;">
</figure>

Comme système de version décentralisé, `git` permet de :

- travailler sur un même projet avec plusieurs personnes, chacun depuis son ordinateur
- réconcilier les historiques pour mettre en commun les modifications effectuées sur différentes _**`branches`**_.

Les _**`branches`**_ peuvent ensuite être fusionnées : 
on parle de _**`merge`**_ ou de _**`rebase`**_.

## Git ≠ GitHub ≠ GitLab

`GitHub` et `GitLab` sont une plateforme en ligne qui utilise Git pour gérer les projets, mais ajoute une couche collaborative et un stockage distant.

<figure style="display: flex; align-items: center; justify-content: center;">
  <img src="media/gitworkflow.png" alt="logo" style="height: 350px; width: auto;">
</figure>



### Git = l'outil (le moteur)

- Logiciel installé localement (sur votre machine)
- Fonctionne en ligne de commande
- Gratuit et open source
- Peut être utilisé seul, sans internet

### GitHub/GitLab = les services (le garage en ligne)

- Plateformes web hébergées
- Stockent vos projets Git en ligne
- Ajoutent des fonctionnalités collaboratives
- Nécessitent une connexion internet

## Propriétés clés 

### Texte 

| Propriété     | Exemple | Description                                                                                                                                  |
| ------------- | ------- | -------------------------------------------------------------------------------------------------------------------------------------------- |
| `color`       | `#333`  | Définit la couleur du texte (ici un gris foncé).                                                                                             |

| `font-weight` | `700`   | Définit l’épaisseur du texte (700 correspond généralement à **bold**).                                                                       |

| Propriété          | Exemple          | Description                                                                              |

## CMS (*Content Management System*) : avantages et inconvénients

### ✓ Les pour
- facile d'utilisation : pas besoin d'ouvrir un éditeur de code, tout passe par des formulaires dans une interface appelée "backend"
- mise en ligne rapide de contenus
- puissant système d'organisation des contenus complexes

### ✗ Les contre
- contraintes du système
    - adapté à des objets standards (blog, site vitrine, e-commerce), les projets métiers spécifiques s'accomodent mal d'un CMS généraliste
    - personnaliser profondément un CMS est complexe, coûteux et fragile

- dépendance
    - multiplication des plugins et des extensions : accumulation de dettes techniques
    - coûts cachés

- inutilement lourd
    - code inutile
    - empilement de couches logicielles
    - ce qui rend une migration coûteuse et complexe

- contenu mal/pas balisé et mal/pas trouvable

</div>

## L'approche SSG (*Static Site Generator*)

### Deux fonctions principales
- convertir d’un format de balisage à un autre, typiquement de Markdown à HTML
- organiser les pages à partir d’instructions écrites dans un langage de sérialisation de données

### Changement de paradigme
- aucun langage dynamique pour servir les pages web
- résultat : un `site statique`

## `Web statique` vs. `Web dynamique` 

### Web statique

- les pages sont écrites en HTML/CSS pur et stockées telles quelles sur le serveur
- pour modifier une page, il faut directement éditer le fichier HTML
- le serveur envoie simplement le fichier au navigateur, sans traitement
- rapide et léger, car pas de calculs côté serveur

![](media/webstatique.png)

## `Web statique` vs. `Web dynamique` 

### Web dynamique

- le contenu est généré "à la volée" par le serveur à chaque demande
- utilise des langages comme PHP, Python, Node.js côté serveur
- s'appuie souvent sur une base de données pour stocker les informations

![](media/webdynamique.png)

## L'approche SSG (*Static Site Generator*)

### ✓ Avantages
- séparation nette entre contenu et fabrique
- création de fichiers HTML hautement structurés
- approche *hands-on* = flexibilité

### ✗ Inconvénients
- difficulté de prise en main

## Exemple de [Quarto](https://quarto.org/about.html)

C'est quoi Quarto ? (*Pandoc sous acid*)

- outil open-source de création et de publication de documents scientifiques (rapports, présentations, livres, sites, etc.)
- permet de combiner du texte (explications, titres, commentaires), du code (R, Python, Julia, etc.), les résultats du code (tableaux, graphiques, calculs)
- production de documents dans différents formats : HTML, PDF, Word, diaporamas, etc. 

### Avantages 

✓ Séparation du fond et de la forme

> Researchers shouldn’t need to choose between LaTeX, MS Word, and HTML but rather be able to author documents that target all of them at the same time. ([Quarto](https://quarto.org/about.html))

✓ Reproductibilté de la recherche

## Un générateur de sites statiques peut interroger Stylo pour constuire un ensemble de pages HTML

- principe de la séparation du fond et de la forme : on stocke le sens, pas la mise en page
