## Éléments de conclusion

Pourquoi des formats plein texte ? 

- considérer le texte comme une donnée
- choisir le balisage => SSP / sortir différents artefacts à partir d'une seule source 
- pérennité / intéropérabilité
- séparation du fond et de la forme

Pourquoi du markdown ? 

- format de balisage léger 
- mais sémantique

Pourquoi un site statique ?

- sobriété de la démarche
- plus facile à héberger et à déployer 
- plus facile à maintenir, plus pérenne
- Les sources sont encodées et facilement récupérable, la migration est facile

Pourquoi utiliser Stylo pour l'écriture et la collaboration ? Quels sont ses avantages et ses inconvénients ?

- collaboration facilitée, en synchrone ou asynchrone
- aperçu rapide du rendu HTML du markdown
- Formulaire pour les métadonnée et la bibliographie pour éviter de manipuler les formats bruts

**Les inconvénients** : on a accès aux templates génériques d'un GSS mais c'est difficile à customiser. Moins adapté à la gestion de contenus complexes. Notamment : où héberger les images ? les contenus non texuels ?

# L'édition de textes anciens

Qu'est-ce qu'un texte ancien ? En connaissez-vous ?

## Que désigne-t-on par "texte ancien" ?

### Monde grec 

- époque archaïque (VIII^e^-VII^e^ s. av. J.-C.) – L'Iliade et l'Odyssée, poèmes d'Hésiode (tradition orale) + poésie lyrique (Sappho, Pindare)
- époque classique (V^e^-IV^e^ s. av. J.-C.) – émergence de nombreux genres
- période hellenistique (III^e^-I^er^ s. av. J.-C.) – poésir érudite (Callimaque ou Théocrite, bibliothèques)

### Monde latin

- assimilation, traduction et renouvellement 
- spectacle (comédies de Plaute ou Térence)
- époque républicaine – Cicéron, Lucrèce, Catulle, Salluste
- règne d'Auguste – Virgile, Tite-Live, Horace ou Ovide, poètes élégiaques
- mais encore Quintilien (grammaire et rhétorique), Tacite (histoire), Pline le Jeune (correspondances) et Pline l’Ancien (sciences et techniques)…

Nous ne possédons pas les autographe… Comment rendre compte de ces textes aujourd'hui ?

## Le support du texte influence 

- les pratiques d’écritures, de lecture, de diffusion (le support est-il disponible, et si oui, à quel prix ? Est-il facile d’y écrire ? Est-il facile à consulter ?)
- la transmission de ces textes au fil des siècles (le document a-t-il résisté au temps et aux intempéries ?)
- les compétences et pratiques d’édition nécessaires pour l’étudier (un papyrologue n’a pas les mêmes pratiques qu’un épigraphiste)

## Que faire de ces textes anciens ?

L’éditeur dispose d’une tradition manuscrite (médiévale), d'un ensemble de `témoins`, de manuscrits transmettant un texte. Ceux-ci, copiés par des copistes différents, dans des lieux et à des époques variées, ne sont pas des copies conformes les uns des autres et comportent des `variantes`.

L'édition crtique veut rétablir l’œuvre d’un auteur dans sa version originale, soit, avant les `altérations` (fautes, lacunes, ajouts, etc.) qui sont le résultat des copies successives. Pour ce faire, l’éditeur doit comparer les différents états conservés dudit texte.

C'est la `démarche philologique`. 

## La philologie

⤿ a une longue histoire, mais se formalise comme discipline au XIX^e^ s.

⤿ différentes école, la méthode fondatrice est celle de `Lachmann` : une enquête génétique basée sur les erreurs

## L'ecdotique

⤿ l’art d’éditer les textes en suivant une méthode critique

⤿ ἐκδίδωμι, « produire au dehors, faire passer en d’autres mains, confier; publier »

> L’édition critique a un rôle de médiation et de restitution de l’univers de création du texte source. (Casenave 2023, 10)

## Édition critique : une approche méthodique et scientifique 

### Une double exigence

> [Cette approche] repose sur la recherche, par-delà la diversité des textes fournis par les multiples manuscrits d’une œuvre donnée parvenue jusqu’à nous, de l’état le plus proche possible de ce que l’on suppose avoir été le texte original – ce qui nécessite de rendre compte de l’histoire de ce texte. (Bureau 2023)

**Donner une version arrêtée du texte mais présenter la diversité et la muliplicité des témoins.** 

## Édition critique : une approche méthodique et scientifique 

### Les étapes de l'édition critique 

🪶 La recensio : on recherche les manuscrits porteurs du texte que l’on veut éditer. 

🪶 La collatio : comparer les différents témoins. Relever les variantes et les compatrer. Identifier les principales familles de manuscrits. Choisir les variantes qui sont pertinentes et doivent être reportées dans l'apparat critique. 

🪶 L’emendatio : l’éditeur corrige les erreurs et corruptions. On dit qu'il choisit la bonne `leçon` ou le `lemme`. 

🪶 La dispositio : mettre en forme l’édition avec les leçons choisies et la constitution de l’apparat critique.

## Écriture vs. saisie de données

> \[L\]e formalisme de la tradition imprimée n’est pas de la même nature que celui de la pratique numérique car les conventions suivies dans les éditions imprimées sont en fait le produit de changements transmis de génération en génération durant plus de trois siècles. La maîtrise de ces conventions s’apparente donc à celle d’une langue pourvue de signes, de termes techniques, de règles et d’ornements de style où le latin occupe la première place. Connaître cette langue, c’est aussi la connaître dans le temps et être en mesure de la suivre à une époque où elle n’était pas codifiée comme elle l’est aujourd’hui par les grandes maisons d’édition. La lecture méthodique de cette langue requiert certes un entraînement particulier, mais elle repose sur l’intelligence humaine et sa faculté d’interpréter naturellement les termes, les signes et l’économie de leur disposition. Et pour la même raison, cette lecture reste inaccessible à une machine. Autrement dit, tandis que les notes critiques des éditions imprimées sont écrites dans une langue suivant une grammaire qui lui est propre, celles des éditions numériques sont saisies dans un code structuré formé d’éléments subordonnés et d’attributs, chacun destiné à recevoir un contenu qui lui est propre. (Alessi 2023)


## Écriture vs. saisie de données

> Aux côtés de la dimension verticale de l’établissement du texte, cela rend en effet possible et réalisable une dimension horizontale d’observation d’une tradition dans la synchronie et l’interrogation de cette synchronie confrontée à d’autres moments de l’histoire textuelle. (Bureau 2023)

## Une troisième voix s'élève (*cf.* Alessi 2023)

- Pas d’opposition stricte entre édition papier et numérique
- Une complémentarité plutôt qu’une rupture
- L’édition critique traditionnelle porte des subtilités interprétatives difficilement traduisibles en balisage
- Le numérique offre une exploration élargie (variantes, parcours de lecture) mais ne remplace pas la finesse de l’apparat savant
- La proposition de Robert Alessi consiste à partir de LaTeX pour sortir un PDF imprimable et un encodage TEI
